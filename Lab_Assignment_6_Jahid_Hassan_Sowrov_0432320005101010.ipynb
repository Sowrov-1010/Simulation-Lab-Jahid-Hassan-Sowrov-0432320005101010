# Chi Square Fit Test
Chi-Square Goodness-of-Fit Test in Python

## Import Necessary Libraries

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
```

## Step 1: Observed data (frequencies in each category)

```python
import numpy as np

# New Student ID last two digits
d_id = 10

# Calculate observed frequencies using specified assignment formulas
obs_c = 20 + d_id
obs_cpp = 25 + (d_id % 10)
obs_java = 30 + (d_id % 7)
obs_python = 35 + (d_id % 5)

observed_freqs = np.array([obs_c, obs_cpp, obs_java, obs_python])
print(f"Observed Frequencies for d={d_id}:", observed_freqs)
```

## Step 2: Expected frequencies (assuming uniform distribution)

```python
total_responses = np.sum(observed_freqs)
expected_freqs = np.full(len(observed_freqs), total_responses / len(observed_freqs))
print(f"Expected Frequencies (Total={total_responses}):", expected_freqs)
```

## Step 3: Chi-square statistic calculation

```python
chi_square_stat = np.sum(((observed_freqs - expected_freqs) ** 2) / expected_freqs)
print(f"Chi-Square Statistic: {chi_square_stat:.5f}")
```

## Step 4: Degrees of freedom

```python
df = len(observed_freqs) - 1
print(f"Degrees of Freedom: {df}")
```

## Step 5: Calculate the p-value and critical value

```python
import scipy.stats as stats

sig_level = 0.05
p_val = 1 - stats.chi2.cdf(chi_square_stat, df)
crit_val = stats.chi2.ppf(1 - sig_level, df)
```

## Display the results

```python
print(f"=== ANALYSIS METRICS ===")
print(f"Calculated Chi-Square Value : {chi_square_stat:.4f}")
print(f"Degrees of Freedom          : {df}")
print(f"Critical Value (alpha=0.05) : {crit_val:.4f}")
print(f"Computed P-Value            : {p_val:.4f}")
```

## Conclusion

```python
if chi_square_stat > crit_val:
    print("Decision: Reject the Null Hypothesis (H0).")
    print("Result: Statistically significant evidence that programming language preferences differ.")
else:
    print("Decision: Fail to Reject the Null Hypothesis (H0).")
    print("Result: No statistically significant difference in programming language preferences.")
```

## Plotting observed vs expected frequencies

```python
import matplotlib.pyplot as plt

categories = ['C Language', 'C++', 'Java Platform', 'Python Language']
plt.figure(figsize=(8, 5))
plt.bar(categories, observed_freqs, color='#3498db', alpha=0.85, label='Observed Data', width=0.35, align='center')
plt.step(categories, expected_freqs, color='#e74c3c', where='mid', label='Expected Uniform Line', linewidth=2.5)
plt.title('Distribution Comparison: Observed vs. Expected', fontsize=12, fontweight='bold')
plt.ylabel('Frequencies Count')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```
